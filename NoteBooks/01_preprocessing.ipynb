```python
import cv2 
import os
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
from PIL import Image, ImageFilter, ImageEnhance
import random
```

```python
IMG_SIZE = (224, 224)
RANDOM_SEED = 42

SPLIT_RATIO = {
    "train": 0.6,
    "val": 0.20,
    "test": 0.20
}

raw_paths = {
    "NOTID": r"D:\Study\ID_Checking\Data\raw\0_NotID",
    "ID": r"D:\Study\ID_Checking\Data\raw\1_ID"
}

processed_paths = {
    "train": {
        "ID": r"D:\Study\ID_Checking\Data\processed\train\ID",
        "NOTID": r"D:\Study\ID_Checking\Data\processed\train\NOTID"
    },
    "val": {
        "ID": r"D:\Study\ID_Checking\Data\processed\val\ID",
        "NOTID": r"D:\Study\ID_Checking\Data\processed\val\NOTID"
    },
    "test": {
        "ID": r"D:\Study\ID_Checking\Data\processed\test\ID",
        "NOTID": r"D:\Study\ID_Checking\Data\processed\test\NOTID"
    }
}

random.seed(RANDOM_SEED)
```

```python
def split_and_save(class_name):
    src_dir = raw_paths[class_name]

    images = [
        f for f in os.listdir(src_dir)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(images)

    total = len(images)
    train_end = int(total * SPLIT_RATIO["train"])
    val_end = train_end + int(total * SPLIT_RATIO["val"])

    splits = {
        "train": images[:train_end],
        "val": images[train_end:val_end],
        "test": images[val_end:]
    }

    for split, files in splits.items():
        dst_dir = processed_paths[split][class_name]
        os.makedirs(dst_dir, exist_ok=True)

        for fname in files:
            src_path = os.path.join(src_dir, fname)
            dst_path = os.path.join(dst_dir, fname)

            try:
                img = Image.open(src_path).convert("RGB")
                img = img.resize(IMG_SIZE)
                img.save(dst_path)
            except Exception as e:
                print("Error:", src_path, e)

    print(
        f"{class_name} | total={total} | "
        f"train={len(splits['train'])} | "
        f"val={len(splits['val'])} | "
        f"test={len(splits['test'])}"
    )

split_and_save("ID")
split_and_save("NOTID")

print("SPLIT + PREPROCESSING DONE CLEAN")    
```
